```python
#Import some libraries for this download
from pathlib import Path
import pandas as pd
import yfinance as yf

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

print("Libraries imported")
```

```python
#Set time frame (Sept 1, 2016 - Sept 1, 2026)

START_DATE = '2016-09-01' #start date is inclusive 
END_DATE = '2026-09-02' #end date is exclusive
```

```python
#Locate our project folder 

current_dir = Path.cwd()

if current_dir.name == 'notebooks':
    project_root = current_dir.parent
else:
    project_root = current_dir
raw_data_dir = project_root / "data" / "raw"

#Confirmation of folders' existence 
print("Project folder:", project_root)
print("Raw data folder:", raw_data_dir)
print("Folder exists?", raw_data_dir.exists())
```

```python
#50 company ticker universe 

tickers = [ 
    #Technology 
    "AAPL","MSFT","NVDA","AVGO","AMD","ORCL","CSCO","CRM","ADBE","INTC",
    #Consumer Staples 
    "WMT","COST","KO","PG",
    #Health Care 
    "LLY","JNJ","UNH","ABBV","PFE",
    #Energy           
    "XOM","CVX","COP",
    #Financials        
    "JPM","BAC","GS","V","BLK",
    #Utilities         
    "NEE","DUK","SO",
    #Consumer Disc.   
    "AMZN","TSLA","HD","MCD","NKE",
    #Materials         
    "LIN","SHW","APD",
    #Industrials    
    "CAT","GE","HON","RTX","UPS",
    #Real Estate      
    "PLD","AMT","O",
    #Communication   
    "GOOGL","META","NFLX","DIS",
]

#check quantity
print("Number of companies:", len(tickers))
```

```python
universe_data = yf.download( tickers=tickers, start= START_DATE, end= END_DATE, 
                           interval="1d", auto_adjust=False, group_by="ticker",
                           actions=False, progress=True, threads=True, timeout=30)

print("Download complete")

print("50-company universe shape:", universe_data.shape)
display(universe_data.head())
```

```python
#Reorganize the universe data into row long format

company_frame = []

for ticker in tickers:
    #handle any potential yfinance column arrangement differences
    if not isinstance(universe_data.columns, pd.MultiIndex):
        if ticker == tickers[0]:
            ticker_df = universe_data.copy()
        else:
            continue

    elif ticker in universe_data.columns.get_level_values(0):
        ticker_df = universe_data[ticker].copy()
        
    elif ticker in univer_data.columns.get_level_values(1):
        ticker_df = universe_data.xs( ticker, axis=1, level=1,).copy()

    else:
        print(f"No columns found for {ticker}")
        continue

    if ticker_df.empty or ticker_df.dropna(how="all").empty:
        print(f"No data available for {ticker}")
        continue 
        
   #Turn date index into normal column
    ticker_df = ticker_df.reset_index()
   #standardize date column name
    ticker_df = ticker_df.rename( columns={ "Date": "Date", "Datetime":"Date"})

    ticker_df["Ticker"] = ticker

    company_frame.append(ticker_df)
```

```python
#Combine all companies 

universe_df = pd.concat(company_frame, ignore_index=True)

#Arrange columns and rows
universe_df["Date"] = pd.to_datetime(universe_df["Date"])
universe_df = universe_df.sort_values( ["Ticker", "Date"]).reset_index(drop=True)

print("Combined data shape:", universe_df.shape)
print("Companies:", universe_df["Ticker"].nunique())

display(universe_df.head(10))
```

```python
#Sanity checks

print("Start date:", universe_df["Date"].min())
print("End date:", universe_df["Date"].max())

print("Rows per company:")
display(universe_df.groupby("Ticker").size().to_frame("Row_Count"))

print("\nMissing values by column:")
display(universe_df.isna().sum().to_frame("Missing_Count"))
```

```python
#save file

universe_file = raw_data_dir / "universe_50.csv"

universe_df.to_csv(universe_file, index=False)

print("Saved to:", universe_file)
print("File exists?", universe_file.exists())
```

```python
#Download AMAZON subset to do preliminary works

amzn_data = yf.download( tickers="AMZN", start=START_DATE, end=END_DATE,
                       interval="1d", auto_adjust=False, actions=False,
                       progress=True, timeout=30)

#mitigating yfinance's potential multiindex columns
if isinstance(amzn_data.columns, pd.MultiIndex):
    
    if "AMZN" in amzn_data.columns.get_level_values(0):
        amzn_data = amzn_data["AMZN"].copy()
        
    elif "AMZN" in amzn_data.columns.get_level_values(1):
        amzn_data = amzn_data.xs("AMZN", axis=1, level=1).copy()
        

#Keep OHLCV, replace Close with adjusted close 
amzn_df = amzn_data[["Open", "High", "Low","Adj Close", "Volume"]].copy()
amzn_df = amzn_df.rename(columns={"Adj Close": "Adj_Close"})

amzn_df.index.name = "Date"
amzn_df = amzn_df.reset_index()

amzn_df = amzn_df[ ["Date", "Open", "High", "Low", "Adj_Close", "Volume"]]

amzn_df["Date"] = pd.to_datetime(amzn_df["Date"])

print("Amazon data shape:", amzn_df.shape)
display(amzn_df.head(10))
```

```python
#Sanity check

print("Date range", amzn_df["Date"].min(), "to", amzn_df["Date"].max())

print("\nMissing values:")
display(amzn_df.isna().sum().to_frame("Missing_Count"))
```

```python
#Save file

amzn_file = raw_data_dir / "amzn.csv"

amzn_df.to_csv(amzn_file, index=False)

print("Saved to:", amzn_file)
print("File exists?", amzn_file.exists())
```
